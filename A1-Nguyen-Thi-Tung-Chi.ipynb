```python
raw_dataset = [
    # Epoch 1
    [1.2, 0.9, None, 1.1, 0.8, 0.7, 18.2, 19.5, 0.6, 0.5, 0.5, 0.4],
    # Epoch 2
    [0.55, 0.52, 0.50, 0.48, 0.47, 0.45, 0.44, 0.43, 0.42, 0.41, 0.40, 0.39],
    # Epoch 3
    [0.39, 0.38, 0.38, 0.39, 0.38, 0.39, 0.38, 0.38, 0.39, 0.38, 0.38, 0.38],
    # Epoch 4
    [0.38, 0.39, 0.38, 0.38, 0.39, 0.38, 0.39, 0.38, 0.38, 0.39, 0.38, 0.39]
]

BATCH_SIZE = 4 
PATIENCE_LIMIT = 2
LR = 0.01
MIN_IMPROVENMENT = 0.01

#Khởi tạo biến theo dõi cho task 5
best_epoch_loss = float ('inf')
patience = 0


print("*** STARTING MULTI-EPOCH TRAINING ENGINE ***\n")

# Task 1: Outer loop
for epoch_index, epoch_data in enumerate(raw_dataset):
    epoch_num = epoch_index +1
    print (f"Epoch {epoch_num}  (LR: {LR:.4f})")

    epoch_loss_sum = 0
    valid_batches_count =0
    batch_num = 1

    #Task 1: Inner loop (Batch slicing)
    for i in range (0,len (epoch_data),BATCH_SIZE):
        batch =epoch_data [i : i + BATCH_SIZE]

        batch_loss_sum = 0
        valid_samples_count = 0

        #Task 2: Sample- Level Filtering
        for sample in batch :
            if sample is None:
                print (f" [Batch {batch_num}] Corrupted sample (None) detected! Skipping sample...")
                continue

            batch_loss_sum += sample
            valid_samples_count += 1
        
        # Nếu toàn bộ batch đều là None (Tránh lỗi chia cho 0)
        if valid_samples_count ==0:
            batch_num +=1
            continue

        #Task 3: Batch Gradient Gủad
        #Tính toán Mean loss của batch 
        batch_mean_loss = batch_loss_sum / valid_samples_count 
        
        if batch_mean_loss > 15.0:
            print (f"[Batch {batch_num}] Exploding Batch Loss ({batch_mean_loss:.4f} > 15.0)! Clipping applied, batch skipped.")
            batch_num += 1
            continue

        # Nếu batch hợp lệ, cộng dồn vào epoch loss
        print (f"[Batch {batch_num}] Process {valid_samples_count} sample. Batch Mean Loss: {batch_mean_loss:.4f}")
        epoch_loss_sum += batch_mean_loss
        valid_batches_count += 1
        batch_num +=1

    
    #Tính Mean Loss của toàn bộ epoch 
    if valid_batches_count > 0:
        epoch_mean_loss = epoch_loss_sum / valid_batches_count 
    else:
        epoch_mean_loss = 0.0

    print (f"Epoch {epoch_num} Mean Loss: {epoch_mean_loss:.4f}\n")



    #Task 4: Sheduler Stub 
    def on_lr_update(new_lr):
        pass


    # Task 5: Cập nhật Learning Rate và Early Stopping
    #Kiểm tra loss có cải thiện ít nhất 0.01 hay khônng

    if epoch_mean_loss <= (best_epoch_loss - MIN_IMPROVENMENT ):
        best_epoch_loss = epoch_mean_loss
        patience = 0 # Reset nếu có cải thiện

    else:
        #Nếu không cải thiện 
        LR = LR *0.5 # Giarm rate đi một nửa
        on_lr_update (LR)
        patience +=1 


        print (f"---> Loss did not improve significantly. LR reduced to {LR:.4f}.")
        print (f"     Patience: {patience}/{PATIENCE_LIMIT}")


        #Kiểm tra xem có bị vướt giới hạn chịu đựng Patience 
        if patience >= PATIENCE_LIMIT:
            print ("\n[EARLY STOPPING TRIGGERED] Aborting training loop.")
            break

print ("*** TRAINING COMPLETED ***")
```
